```python
# Mini-Projeto Avaliativo: Análise de Dados com Python - dados de varejo

import pandas as pd
import numpy as np
from IPython.display import display
# Ler o arquivo usando o separador correto
df = pd.read_csv("Base Varejo.csv", sep=";")

# Verificar a estrutura original
print("Dimensões da base:", df.shape)

print("\nColunas identificadas:")
print(df.columns.tolist())

print("\nInformações da base:")
df.info()

print("\nPrimeiros registos:")
display(df.head())
```

```python
#DIAGNÓSTICO E LIMPEZA INICIAL

# Identificar as colunas totalmente vazias
colunas_vazias = df.columns[df.isna().all()].tolist()
print("Colunas totalmente vazias:", colunas_vazias)

# Remover apenas as colunas totalmente vazias
df = df.dropna(axis=1, how="all")
print("\nDimensões após a limpeza:", df.shape)

print("\nColunas restantes:")
print(df.columns.tolist())

df.info()
```

```python
#TRANSFORMAÇÃO E VALIDAÇÃO

# Verificar o formato atual da coluna DATA
print("Tipo original da coluna DATA:", df["DATA"].dtype)
print("\nExemplos de datas:")
print(df["DATA"].head())

# Converter DATA para o formato datetime
df["DATA"] = pd.to_datetime(
    df["DATA"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Verificar se existem datas inválidas
datas_invalidas = df["DATA"].isna().sum()
print("\nDatas inválidas após a conversão:", datas_invalidas)
print("Tipo após a conversão:", df["DATA"].dtype)

# Verificar valores nulos por coluna
print("\nValores nulos por coluna:")
print(df.isna().sum())

# Investigar categorias e nomes de produtos vazios
categorias_vazias = (
    df["PR_CAT"].astype("string").str.strip().eq("").sum()
)

nomes_vazios = (
    df["PR_NOME"].astype("string").str.strip().eq("").sum()
)

print("\nCategorias de produtos vazias:", categorias_vazias)
print("Nomes de produtos vazios:", nomes_vazios)

# Verificar registos totalmente duplicados
duplicados = df.duplicated().sum()
print("\nRegistos totalmente duplicados:", duplicados)

# Confirmar a estrutura atual da base
print("\nDimensões atuais da base:", df.shape)
print("\nTipos de dados atuais:")
print(df.dtypes)
```

```python
#INVESTIGAÇÃO DE DUPLICADOS

# Contar as linhas que fazem parte de duplicações exatas
mascara_duplicados = df.duplicated(keep=False)
linhas_duplicadas = df[mascara_duplicados]

print("Total de linhas envolvidas em duplicações:",
      len(linhas_duplicadas))

print("\nExemplos de registos duplicados:")
display(linhas_duplicadas.head(10))

# Verificar quantas linhas existem por combinação de compra e produto
repeticoes_compra_produto = df.groupby(
    ["CO_ID", "PR_ID"]
).size()

print("\nCombinações compra-produto repetidas:",
      (repeticoes_compra_produto > 1).sum())

print("\nCombinações compra-produto com mais de uma linha:")
display(
    repeticoes_compra_produto[
        repeticoes_compra_produto > 1
    ].head(10)
)
```

```python
# Validar a consistência dos identificadores de compra

validacao_compras = df.groupby("CO_ID").agg(
    quantidade_clientes=("CL_ID", "nunique"),
    quantidade_datas=("DATA", "nunique"),
    quantidade_linhas=("CO_ID", "size")
)

print("Número de compras distintas:", df["CO_ID"].nunique())

print("Compras associadas a vários clientes:",
      (validacao_compras["quantidade_clientes"] > 1).sum())

print("Compras associadas a várias datas:",
      (validacao_compras["quantidade_datas"] > 1).sum())

print("\nExemplos de compras com vários clientes:")
display(
    validacao_compras[
        validacao_compras["quantidade_clientes"] > 1
    ].head(10)
)
```

```python
#TRATAMENTO DE DUPLICADOS EXATOS

# Guardar uma cópia antes da remoção, para podermos comparar
df_antes_deduplicacao = df.copy()

# Contar duplicados exatos antes do tratamento
duplicados_antes = df.duplicated().sum()

# Remover duplicados exatos, mantendo a primeira ocorrência
df = df.drop_duplicates().copy()

# Verificar o resultado
duplicados_depois = df.duplicated().sum()

print("=== RESULTADO DA LIMPEZA ===")
print("Registos antes:", len(df_antes_deduplicacao))
print("Duplicados exatos identificados:", duplicados_antes)
print("Registos removidos:", len(df_antes_deduplicacao) - len(df))
print("Registos após a limpeza:", len(df))
print("Duplicados restantes:", duplicados_depois)
print("Dimensões finais desta etapa:", df.shape)
```

```python
# NORMALIZAÇÃO DOS CAMPOS DE TEXTO

colunas_texto = ["CL_GENERO", "CL_SEG", "PR_CAT", "PR_NOME"]

for coluna in colunas_texto:
    # Converter para texto, remover espaços externos,
    # reduzir espaços consecutivos e padronizar maiúsculas
    df[coluna] = (
        df[coluna]
        .astype("string")
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
        .str.upper()
    )

print("Campos de texto normalizados.")

print("\nExemplos após a normalização:")
display(df[colunas_texto].head())

print("\nValores nulos após a normalização:")
print(df[colunas_texto].isna().sum())
```

```python
# REGRAS DE VALIDAÇÃO

# Verificar identificadores ausentes
colunas_id = ["CO_ID", "CL_ID", "PR_ID"]

print("Identificadores nulos:")
print(df[colunas_id].isna().sum())

# Verificar compras associadas a mais de um cliente
clientes_por_compra = df.groupby("CO_ID")["CL_ID"].nunique()
compras_multiplos_clientes = (clientes_por_compra > 1).sum()

# Verificar compras associadas a mais de uma data
datas_por_compra = df.groupby("CO_ID")["DATA"].nunique()
compras_multiplas_datas = (datas_por_compra > 1).sum()

# Verificar validade dos campos numéricos
print("\nValores distintos de CL_GENERO:")
print(df["CL_GENERO"].value_counts(dropna=False))

print("\nValores distintos de CL_EC:")
print(df["CL_EC"].value_counts(dropna=False).sort_index())

print("\nValores distintos de CL_FHL:")
print(df["CL_FHL"].value_counts(dropna=False).sort_index())

print("\nRESULTADOS DAS REGRAS DE VALIDAÇÃO")
print(
    "Compras associadas a mais de um cliente:",
    compras_multiplos_clientes
)
print(
    "Compras associadas a mais de uma data:",
    compras_multiplas_datas
)
print(
    "Datas inválidas ou ausentes:",
    df["DATA"].isna().sum()
)

# Validar se o número de filhos está dentro do intervalo observado
filhos_invalidos = (
    df["CL_FHL"].isna()
    | ~df["CL_FHL"].between(0, 4)
)

print(
    "Registros com CL_FHL ausente ou fora do intervalo de 0 a 4:",
    filhos_invalidos.sum()
)
```

```python
# ESTATÍSTICA DESCRITIVA DE CL_FHL


filhos = df["CL_FHL"].dropna()

estatisticas_filhos = pd.Series({
    "Quantidade de registros válidos": filhos.count(),
    "Média": filhos.mean(),
    "Mediana": filhos.median(),
    "Desvio-padrão": filhos.std(),
    "Moda": filhos.mode().iloc[0] if not filhos.mode().empty else np.nan,
    "Valor mínimo": filhos.min(),
    "Primeiro quartil (25%)": filhos.quantile(0.25),
    "Segundo quartil (50%)": filhos.quantile(0.50),
    "Terceiro quartil (75%)": filhos.quantile(0.75),
    "Valor máximo": filhos.max()
})

print("ESTATÍSTICA DESCRITIVA — NÚMERO DE FILHOS")
display(estatisticas_filhos.to_frame(name="Valor").round(2))

print("\nDistribuição do número de filhos:")
display(
    df["CL_FHL"]
    .value_counts(dropna=False)
    .sort_index()
    .rename_axis("Número de filhos")
    .to_frame(name="Quantidade de registros")
)
```

```python
# COMPRAS POR GÊNERO

compras_por_genero = (
    df.groupby("CL_GENERO")
    .agg(
        registros=("CO_ID", "size"),
        compras_distintas=("CO_ID", "nunique"),
        clientes_distintos=("CL_ID", "nunique")
    )
    .sort_values("compras_distintas", ascending=False)
)

print("ANÁLISE POR GÊNERO")
display(compras_por_genero)
```

```python
# DISTRIBUIÇÃO POR CATEGORIA

produtos_por_categoria = (
    df.groupby("PR_CAT")
    .agg(
        registros_de_produtos=("PR_ID", "size"),
        produtos_distintos=("PR_ID", "nunique"),
        compras_com_categoria=("CO_ID", "nunique")
    )
    .sort_values("registros_de_produtos", ascending=False)
)

print("ANÁLISE POR CATEGORIA")
display(produtos_por_categoria)
```

```python
# TABELA DINÂMICA: GÊNERO × SEGMENTO

tabela_genero_segmento = pd.pivot_table(
    df,
    index="CL_GENERO",
    columns="CL_SEG",
    values="CO_ID",
    aggfunc="nunique",
    fill_value=0
)

print("COMPRAS DISTINTAS POR GÊNERO E SEGMENTO")
display(tabela_genero_segmento)
```
